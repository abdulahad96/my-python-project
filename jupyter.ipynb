````python
# Data Cleaning Functions
print("🧹 Data Cleaning Implementation:")

def clean_tutor_data(df):
    """
    Clean and standardize the tutor data
    """
    df_clean = df.copy()
    
    # 1. Clean city names
    df_clean['city'] = df_clean['city'].str.strip().str.title()
    
    # 2. Clean years_experience
    def extract_years(exp_str):
        if pd.isna(exp_str):
            return None
        exp_str = str(exp_str).lower().strip()
        
        # Handle text values
        if 'year' in exp_str:
            return int(exp_str.split()[0])
        elif exp_str in ['one', '1']:
            return 1
        elif exp_str in ['two', '2']:
            return 2
        elif exp_str in ['three', '3']:
            return 3
        elif exp_str in ['four', '4']:
            return 4
        elif exp_str in ['five', '5']:
            return 5
        elif exp_str in ['six', '6']:
            return 6
        elif exp_str in ['seven', '7']:
            return 7
        elif exp_str in ['eight', '8']:
            return 8
        elif exp_str in ['nine', '9']:
            return 9
        elif exp_str in ['ten', '10']:
            return 10
        elif exp_str == '—':
            return None
        else:
            try:
                return int(exp_str)
            except:
                return None
    
    df_clean['years_experience_clean'] = df_clean['years_experience'].apply(extract_years)
    
    # 3. Clean hourly_rate
    def clean_hourly_rate(rate_str):
        if pd.isna(rate_str) or str(rate_str).strip() in ['N/A', '']:
            return None
        
        rate_str = str(rate_str).strip()
        
        # Remove currency symbols and convert to PKR
        if 'PKR' in rate_str:
            return float(rate_str.replace('PKR', '').replace(',', '').strip())
        elif '$' in rate_str:
            # Convert USD to PKR (approximate rate: 1 USD = 280 PKR)
            usd_amount = float(rate_str.replace('$', '').replace(',', '').strip())
            return usd_amount * 280
        elif 'usd' in rate_str.lower():
            usd_amount = float(rate_str.lower().replace('usd', '').replace(',', '').strip())
            return usd_amount * 280
        else:
            try:
                return float(rate_str.replace(',', '').strip())
            except:
                return None
    
    df_clean['hourly_rate_clean'] = df_clean['hourly_rate'].apply(clean_hourly_rate)
    
    # 4. Clean subjects
    df_clean['subjects_clean'] = df_clean['subjects'].str.strip().str.title()
    
    # 5. Clean rating (ensure it's between 1-5)
    df_clean['rating_clean'] = df_clean['rating'].clip(lower=1.0, upper=5.0)
    
    return df_clean

# Apply cleaning
df_cleaned = clean_tutor_data(df)

print_llm_response({
    "Original Shape": df.shape,
    "Cleaned Shape": df_cleaned.shape,
    "New Columns Added": ["years_experience_clean", "hourly_rate_clean", "subjects_clean", "rating_clean"]
}, "Data Cleaning Results")
````

````python
# Data Cleaning using pandas methods (as suggested)
print("🧹 Data Cleaning with pandas methods:")

# Create a copy for cleaning
df_clean = df.copy()

# 1. Clean years_experience using pd.to_numeric with errors='coerce'
print("1. Cleaning years_experience...")
# First, handle text values manually
def clean_years(exp):
    if pd.isna(exp):
        return exp
    exp_str = str(exp).lower().strip()
    
    # Handle common text patterns
    text_to_num = {
        'one': 1, 'two': 2, 'three': 3, 'four': 4, 'five': 5,
        'six': 6, 'seven': 7, 'eight': 8, 'nine': 9, 'ten': 10,
        '—': np.nan
    }
    
    if exp_str in text_to_num:
        return text_to_num[exp_str]
    
    # Handle "X+" patterns
    if '+' in exp_str:
        return int(exp_str.replace('+', ''))
    
    # Handle "X years" patterns
    if 'year' in exp_str:
        return int(exp_str.split()[0])
    
    return exp

df_clean['years_experience'] = df_clean['years_experience'].apply(clean_years)
df_clean['years_experience'] = pd.to_numeric(df_clean['years_experience'], errors='coerce')

print(f"Years experience cleaned. Data type: {df_clean['years_experience'].dtype}")
print(f"Non-null count: {df_clean['years_experience'].count()}")
````

````python
# 2. Clean hourly_rate using regex to strip symbols
print("\n2. Cleaning hourly_rate...")

def clean_hourly_rate(rate):
    if pd.isna(rate) or str(rate).strip() in ['N/A', '']:
        return np.nan
    
    rate_str = str(rate).strip()
    
    # Remove currency symbols using regex
    rate_str = re.sub(r'[PKR$,\s]', '', rate_str)
    
    # Handle "usd" suffix
    if 'usd' in rate_str.lower():
        rate_str = rate_str.lower().replace('usd', '')
        # Convert USD to PKR (1 USD = 280 PKR)
        try:
            return float(rate_str) * 280
        except:
            return np.nan
    
    # Convert to numeric
    try:
        return float(rate_str)
    except:
        return np.nan

df_clean['hourly_rate'] = df_clean['hourly_rate'].apply(clean_hourly_rate)
df_clean['hourly_rate'] = pd.to_numeric(df_clean['hourly_rate'], errors='coerce')

print(f"Hourly rate cleaned. Data type: {df_clean['hourly_rate'].dtype}")
print(f"Non-null count: {df_clean['hourly_rate'].count()}")
print(f"Sample cleaned rates: {df_clean['hourly_rate'].dropna().head().tolist()}")
````

````python
# 3. Clean rating column
print("\n3. Cleaning rating...")

# Clip ratings to valid range (1-5)
df_clean['rating'] = df_clean['rating'].clip(lower=1.0, upper=5.0)
df_clean['rating'] = pd.to_numeric(df_clean['rating'], errors='coerce')

print(f"Rating cleaned. Data type: {df_clean['rating'].dtype}")
print(f"Non-null count: {df_clean['rating'].count()}")
print(f"Rating range: {df_clean['rating'].min()} - {df_clean['rating'].max()}")

# 4. Clean subjects using .str.strip().str.lower()
print("\n4. Cleaning subjects...")

# Normalize subjects: strip spaces and convert to lowercase
df_clean['subjects'] = df_clean['subjects'].str.strip().str.lower()

print(f"Subjects cleaned. Sample:")
print_llm_response(df_clean['subjects'].dropna().head(10).tolist(), "Cleaned Subjects Sample")

# 5. Clean city names
print("\n5. Cleaning city names...")
df_clean['city'] = df_clean['city'].str.strip().str.title()

print(f"City names cleaned. Sample:")
print_llm_response(df_clean['city'].value_counts().head().to_dict(), "Cleaned City Names")
````

````python
# Final comparison and summary
print("📊 Final Data Quality Summary:")

# Compare original vs cleaned
comparison = {
    "Original Shape": df.shape,
    "Cleaned Shape": df_clean.shape,
    "Data Types Fixed": {
        "years_experience": f"{df['years_experience'].dtype} → {df_clean['years_experience'].dtype}",
        "hourly_rate": f"{df['hourly_rate'].dtype} → {df_clean['hourly_rate'].dtype}",
        "rating": f"{df['rating'].dtype} → {df_clean['rating'].dtype}"
    },
    "Missing Values (Original)": df[['years_experience', 'hourly_rate', 'rating']].isnull().sum().to_dict(),
    "Missing Values (Cleaned)": df_clean[['years_experience', 'hourly_rate', 'rating']].isnull().sum().to_dict()
}

print_llm_response(comparison, "Before vs After Cleaning")

# Show cleaned data sample
print("\n✅ Sample of Cleaned Data:")
sample_cleaned = df_clean[['tutor_id', 'name', 'city', 'years_experience', 'hourly_rate', 'rating', 'subjects']].head()
print(sample_cleaned)

# Save cleaned data
df_clean.to_csv('data_cleaned_pandas.csv', index=False)
print_llm_response("Cleaned data saved to 'data_cleaned_pandas.csv'", "File Saved Successfully")
````
